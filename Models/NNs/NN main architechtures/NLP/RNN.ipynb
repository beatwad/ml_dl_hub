### Key idea of LSTM



Cell state moves through all blocks


Every cell has 4 gates (and 4 matrices of N x N size concatenated into matrice of 4*N x N size):

### Forget gate


- if 1 -> transmit block state completely
- if 0 -> forget previous block state

### Input gate



Defines what part of new information is taken in account in the block state

### Update gate



new_state = (old_state | gate) + (processed_state | gate)

### Output gate



Creates output and hidden state

# Quasi RNN



Use recurrent (dynamic) average pooling in CNN (only pooling layer is recurrent in this net).

16 times faster than LSTM.

Not so powerful as classic CNN (can't build too deep CNN with dynamic average pool) but can be used for some tasks where CNN and RNN mix can be useful.

# Echo State Networks (ESN) and Liquid State Machines (LSM)



Main idea of ESN: set dynamic reservoir - part of network with recurrent structure that is not trained (marked with grey color) and train only not recurrent part of network using values from dynamic reservoir (echoes)

Mian idea of LSM: same but use impulse neurons with binary input.
